```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, CharacterTextSplitter
```

```python
# 1. Simulamos un texto técnico largo (o podrías cargar un PDF con pypdf)
texto_tecnico = """
MANUAL DE MANTENIMIENTO - MOTOR V6-TURBO
1. INTRODUCCIÓN
El motor V6-Turbo requiere un mantenimiento preventivo cada 10.000km. 
Es vital revisar el nivel de aceite. El código de error E-404 indica 
una presión baja en el sistema de inyección.

2. PROCEDIMIENTO DE EMERGENCIA
Si observa humo denso, apague el motor inmediatamente. 
El protocolo de seguridad S-99 requiere el uso de guantes térmicos. 
No intente abrir la válvula de escape si la temperatura supera los 90 grados.

3. ESPECIFICACIONES TÉCNICAS
- Presión nominal: 2.5 bar.
- Torque máximo: 450 Nm.
- Capacidad de refrigerante: 5 litros.
"""
```

```python
# 2. Configuramos el Chunking Naïve (Corta por caracteres exactos)
naive_splitter = CharacterTextSplitter(
    separator="", 
    chunk_size=100, 
    chunk_overlap=0
)
```

```python
# 3. Configuramos el Recursive Character Splitting (El estándar de la industria)
recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,      # Tamaño objetivo
    chunk_overlap=20,    # Solapamiento para no perder contexto
    separators=["\n\n", "\n", " ", ""] # Jerarquía de corte
)
```

```python
# 4. Ejecutamos ambos
chunks_naive = naive_splitter.split_text(texto_tecnico)
chunks_recursive = recursive_splitter.split_text(texto_tecnico)
```

```python
print(f"--- RESULTADO NAÏVE ({len(chunks_naive)} chunks) ---")
for i, chunk in enumerate(chunks_naive[:4]):
    print(f"Chunk {i+1}: '{chunk}'")
```

```python
print(f"--- RESULTADO RECURSIVO ({len(chunks_recursive)} chunks) ---")
for i, chunk in enumerate(chunks_recursive[:4]):
    print(f"Chunk {i+1}: '{chunk}'")
```
