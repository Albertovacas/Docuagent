```python
from sentence_transformers import SentenceTransformer, util
import pandas as pd
```

```python
# 1. Cargamos el modelo (Se descargará la primera vez)
# Usamos uno ligero pero potente para empezar
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# 2. El "Manual" de nuestro DocuAgent (Mini-base de conocimientos)
manual_chunks = [
    "Para configurar la base de datos, edite el archivo config.yaml.",
    "El Supervisor Agent coordina a los sub-agentes usando el patrón A2A.",
    "Los errores de autenticación devuelven un código HTTP 401.",
    "Qdrant se utiliza como base de datos vectorial para almacenar embeddings.",
    "La ingesta de PDFs se realiza mediante un pipeline de OCR en GCP."
]
```

```python
# 3. Generamos los embeddings de nuestro manual
manual_embeddings = model.encode(manual_chunks)
```

```python
# 4. Simulación de una pregunta del usuario
query = "¿Cómo se conectan los agentes entre sí?"
query_embedding = model.encode(query)
```

```python
# 5. Cálculo de similitud (The Retrieval Step)
# Comparamos la pregunta contra TODOS los fragmentos del manual
hits = util.semantic_search(query_embedding, manual_embeddings, top_k=2)
```

```python
print(f"Pregunta del usuario: {query}\n")
print("Resultados encontrados:")
for hit in hits[0]:
    texto = manual_chunks[hit['corpus_id']]
    score = hit['score']
    print(f"- [{score:.4f}] {texto}")
```
