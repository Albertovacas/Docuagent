```python
pdf_path = os.getcwd() + "/../data/DeepSeek_V4.pdf"
reader = PdfReader(pdf_path)
total_pages = len(reader.pages)
```

```python
hybrid_parser = HybridParser()
```

```python
print(f"🚀 Iniciando orquestador híbrido: {total_pages} páginas a procesar.")

for i in range(total_pages):
    page_num = i + 1
    page = reader.pages[i]
    raw_text = page.extract_text()
    
    if hybrid_parser._is_complex(raw_text):
        print(f"  [Pág {page_num}/{total_pages}] ✨ MODO VISIÓN: Procesando estructura compleja...")
    else:
        print(f"  [Pág {page_num}/{total_pages}] 📝 MODO TEXTO: Extracción rápida.")
```
