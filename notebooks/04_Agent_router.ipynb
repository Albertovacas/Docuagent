```python
from pydantic import BaseModel, Field
from typing import Literal
```

```python
# 1. Definimos las opciones del Router
class RouteDecision(BaseModel):
    tool_to_use: Literal["manual_rag", "web_search", "direct_answer"] = Field(
        description="La herramienta que mejor responde a la duda del usuario"
    )
    reasoning: str = Field(description="Breve explicación de por qué eligió esa herramienta")
```

```python
# 2. Simulamos la lógica del Supervisor
def supervisor_router(query: str):
    print(f"DEBUG: Analizando query: '{query}'")
    
    # Aquí en un caso real llamaríamos al LLM pidiendo este esquema JSON
    # Vamos a simular la lógica que seguiría el LLM:
    if "precio" in query.lower() or "coste" in query.lower():
        return RouteDecision(tool_to_use="web_search", reasoning="El usuario pregunta por costes actuales.")
    elif "error" in query.lower() or "manual" in query.lower():
        return RouteDecision(tool_to_use="manual_rag", reasoning="Es una duda técnica contenida en el manual.")
    else:
        return RouteDecision(tool_to_use="direct_answer", reasoning="Es un saludo o una duda general.")
```

```python
# Pruebas de ejecución
queries = [
    "¿Cómo arreglo el fallo de inyección?",
    "¿Cuánto vale el repuesto de la turbina?"
]
```

```python
for q in queries:
    decision = supervisor_router(q)
    print(f"Decisión: {decision.tool_to_use} | Razón: {decision.reasoning}\n")
```
